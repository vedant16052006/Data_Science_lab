# Factorial


### Factorial Function

This code defines a function `factorial(n)` that calculates the factorial of a non-negative integer `n`. The factorial of a non-negative integer `n` is the product of all positive integers less than or equal to `n`.

```python
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        fact = 1
        for i in range(1, n + 1):
            fact *= i
        return fact

number = 5
result = factorial(number)
print(f"The factorial of {number} is {result}")
```

# Odd and Even No


### Odd or Even Number

This code defines a function `check_odd_even(n)` that determines if a given integer `n` is odd or even. It uses the modulo operator (`%`) to check if the number has a remainder when divided by 2.

```python
def check_odd_even(number):
    if number % 2 == 0:
        return "Even"
    else:
        return "Odd"

# Example usage:
num1 = 7
print(f"The number {num1} is {check_odd_even(num1)}.")

num2 = 12
print(f"The number {num2} is {check_odd_even(num2)}.")
```

# Palindrome No

### Palindrome Number Check

This code defines a function `is_palindrome(n)` that checks if a given number `n` is a palindrome. A number is considered a palindrome if it reads the same forwards and backwards. The function converts the number to a string and then compares it with its reversed version.

```python
def is_palindrome(number):
    # Convert the number to a string
    num_str = str(number)
    # Check if the string is equal to its reverse
    if num_str == num_str[::-1]:
        return True
    else:
        return False

# Example usage with the number 16:
num = 16
if is_palindrome(num):
    print(f"The number {num} is a palindrome.")
else:
    print(f"The number {num} is not a palindrome.")

# Another example:
num_palindrome = 121
if is_palindrome(num_palindrome):
    print(f"The number {num_palindrome} is a palindrome.")
else:
    print(f"The number {num_palindrome} is not a palindrome.")
```

## Introduction to Expert Systems

An **Expert System** is a computer program that simulates the judgment and behavior of a human or an organization that has expert knowledge and experience in a particular domain. Developed as a prominent branch of Artificial Intelligence (AI) in the 1970s and 1980s, their primary purpose is to solve complex problems and make decisions by reasoning through bodies of knowledge, represented primarily as if–then rules rather than through conventional procedural code.

### History of Expert Systems
The concept of expert systems emerged from AI research at Stanford University in the 1960s. Early notable systems included DENDRAL (for chemical structure analysis) and MYCIN (for diagnosing infectious diseases). These systems demonstrated the potential of AI to capture human expertise and apply it to practical problems, paving the way for further development in various fields.

### Core Components of an Expert System

1.  **Knowledge Base (KB)**:
    *   **What it is**: The heart of an expert system, containing domain-specific facts, heuristics (rules of thumb), and rules. This knowledge is typically acquired from human experts in the field.
    *   **How it works**: It stores structured information, often in the form of 'IF-THEN' rules (e.g., "IF a patient has a fever AND a cough, THEN suspect a cold") or frames (objects with attributes).
    *   **Example**: In a medical diagnostic system, the knowledge base would contain symptoms, diseases, and rules linking them.

2.  **Inference Engine (IE)**:
    *   **What it is**: The 'brain' of the expert system. It is a program that uses logical rules to process the information in the knowledge base and draw conclusions or make recommendations.
    *   **How it works**: It employs reasoning strategies such as:
        *   **Forward Chaining**: Starts with known facts and applies rules to deduce new facts until a goal is reached (data-driven).
        *   **Backward Chaining**: Starts with a goal and works backward to find the facts that would prove or disprove it (goal-driven).
    *   **Example**: Given a patient's symptoms, the inference engine would use the rules in the knowledge base to arrive at a diagnosis.

3.  **User Interface (UI)**:
    *   **What it is**: The component that allows the user to interact with the expert system. It facilitates inputting data (e.g., symptoms, problems) and receiving output (e.g., diagnosis, solution, explanation).
    *   **How it works**: It should be user-friendly, allowing non-experts to effectively use the system. Often, it also provides an explanation facility, which can justify the system's reasoning and conclusions, building user trust.
    *   **Example**: A chatbot-like interface where a user can enter symptoms and receive potential diagnoses and explanations.

### How Expert Systems Aid in Decision-Making
Expert systems aid in decision-making by:
*   **Mimicking human expertise**: They capture and preserve the knowledge of highly skilled individuals, making it accessible to a wider audience.
*   **Consistency**: They apply rules consistently, eliminating human biases and ensuring uniform decision-making.
*   **Speed and Efficiency**: They can process large amounts of information and arrive at conclusions much faster than humans, especially in complex scenarios.
*   **Accessibility**: They make expert-level knowledge available on demand, even when human experts are scarce or unavailable.
*   **Training**: They can serve as valuable training tools for less experienced personnel by explaining their reasoning process.
*   **Risk Reduction**: By providing well-reasoned recommendations, they can help reduce errors and improve the quality of decisions in critical domains like medicine, finance, and engineering.

### Problem Domain: Activity Suggestion System

#### Objective:
The objective of this expert system is to suggest an appropriate activity to the user based on the current weather conditions (sunny, rainy, cloudy) and temperature (hot, mild, cold). This system aims to provide simple, actionable recommendations for outdoor or indoor activities.

#### Defined 'If-Then' Rules:

*   **Rule 1:** IF weather is 'sunny' AND temperature is 'hot' THEN suggest 'Go swimming' or 'Stay indoors with AC'.
*   **Rule 2:** IF weather is 'sunny' AND temperature is 'mild' THEN suggest 'Go for a walk' or 'Have a picnic'.
*   **Rule 3:** IF weather is 'sunny' AND temperature is 'cold' THEN suggest 'Go for a brisk walk' or 'Enjoy hot cocoa outdoors'.
*   **Rule 4:** IF weather is 'rainy' THEN suggest 'Read a book', 'Watch a movie', or 'Play board games'.
*   **Rule 5:** IF weather is 'cloudy' AND temperature is 'mild' THEN suggest 'Visit a museum' or 'Go shopping'.
*   **Rule 6:** IF weather is 'cloudy' AND temperature is 'hot' THEN suggest 'Go to an indoor mall' or 'Visit an art gallery'.
*   **Rule 7:** IF weather is 'cloudy' AND temperature is 'cold' THEN suggest 'Visit a coffee shop' or 'Do some indoor crafting'.

## Implement Expert System Logic

### Subtask:
Write Python code to implement the expert system, including creating a structure for the knowledge base (the rules) and an inference engine that takes user input, applies the rules, and provides a decision.


**Reasoning**:
The subtask requires implementing the expert system logic by defining a knowledge base with rules and an inference engine function to process user input against these rules. This code block will create the knowledge base and the inference engine as specified.



```python
knowledge_base = [
    {
        "conditions": {
            "weather": "sunny",
            "temperature": "warm"
        },
        "suggestions": ["Go for a walk", "Have a picnic", "Play outdoor games"]
    },
    {
        "conditions": {
            "weather": "sunny",
            "temperature": "cold"
        },
        "suggestions": ["Go for a brisk walk", "Visit an outdoor market"]
    },
    {
        "conditions": {
            "weather": "rainy",
            "temperature": "any"
        },
        "suggestions": ["Read a book", "Watch a movie", "Play board games"]
    },
    {
        "conditions": {
            "weather": "cloudy",
            "temperature": "warm"
        },
        "suggestions": ["Visit a museum", "Go to a cafe"]
    },
    {
        "conditions": {
            "weather": "cloudy",
            "temperature": "cold"
        },
        "suggestions": ["Visit an art gallery", "Try an indoor sport"]
    }
]

def inference_engine(user_weather, user_temperature):
    for rule in knowledge_base:
        match_weather = False
        match_temperature = False

        # Check weather condition
        if rule['conditions']['weather'] == user_weather.lower():
            match_weather = True

        # Check temperature condition, handle 'any' case
        if rule['conditions']['temperature'] == "any" or rule['conditions']['temperature'] == user_temperature.lower():
            match_temperature = True

        if match_weather and match_temperature:
            return rule['suggestions']

    return ["Sorry, I can't suggest an activity for these conditions."]

print("Knowledge base and inference engine defined.")
```

## Demonstrate and Test System

### Subtask:
Provide example inputs to the implemented expert system to demonstrate how it processes information and makes decisions based on the predefined rules, and print the resulting recommendations or conclusions.


**Reasoning**:
The subtask requires demonstrating the expert system by providing example inputs to the `inference_engine` function and printing the resulting recommendations. This code block will define example inputs, iterate through them, call the inference engine, and print the output clearly.



```python
print("\n--- Demonstrating Expert System with Examples ---")

# 1. Create a list of example inputs
example_inputs = [
    ("sunny", "warm"),
    ("rainy", "cold"),
    ("cloudy", "mild"), # Note: 'mild' might not match 'warm' directly in the KB, but let's test it.
    ("snowy", "hot"), # This should trigger the 'no suggestion' case
    ("sunny", "cold"),
    ("cloudy", "warm"),
    ("rainy", "any")
]

# 2. Iterate through the list of example inputs
for weather, temperature in example_inputs:
    # 3. Call the inference_engine function
    suggestions = inference_engine(weather, temperature)

    # 4. Print the input and suggestions
    print(f"\nInput: Weather = {weather.capitalize()}, Temperature = {temperature.capitalize()}")
    print(f"Suggestions: {', '.join(suggestions)}")

print("\n--- Demonstration Complete ---")
```
